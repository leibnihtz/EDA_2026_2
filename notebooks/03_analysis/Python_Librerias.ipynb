```python
import time
import sys
import pandas as pd
import numpy as np

# 1. GENERACIÓN DE DATOS (1 Millón de registros)
N = 10000000
np.random.seed(42)

# Datos para Python Nativo (Lista de Diccionarios)
edades_list = np.random.randint(18, 65, size=N).tolist()
salarios_list = np.random.uniform(1000, 5000, size=N).tolist()
datos_python = [{"edad": e, "salario": s} for e, s in zip(edades_list, salarios_list)]

# Datos para Pandas (DataFrame)
df = pd.DataFrame({"edad": edades_list, "salario": salarios_list})


print("=== 1. COMPARACIÓN DE MEMORIA RAM ===")

# Memoria de la lista contenedora + estimación real de los diccionarios internos
mem_lista_contenedor = sys.getsizeof(datos_python)
mem_muestra_dicts = sum(
    sys.getsizeof(d) + sys.getsizeof(d["edad"]) + sys.getsizeof(d["salario"]) 
    for d in datos_python[:1000]
)
mem_python_total = mem_lista_contenedor + (mem_muestra_dicts / 1000) * N
mem_pandas_total = df.memory_usage(deep=True).sum()

print(f"Python Nativo (Lista de dicts): ~{mem_python_total / (1024**2):.2f} MB")
print(f"Pandas (DataFrame):             ~{mem_pandas_total / (1024**2):.2f} MB")
print(f"Pandas ocupa ~{mem_python_total / mem_pandas_total:.1f} veces menos memoria RAM.")


print("\n=== 2. COMPARACIÓN (Filtrar edad > 30 y calcular promedio) ===")

# --- Prueba con Python Nativo ---
inicio = time.time()
filtrados = [row["salario"] for row in datos_python if row["edad"] > 30]
promedio_python = sum(filtrados) / len(filtrados)
tiempo_python = (time.time() - inicio) * 1000  # Convertido a milisegundos

# --- Prueba con Pandas ---
inicio = time.time()
promedio_pandas = df[df["edad"] > 30]["salario"].mean()
tiempo_pandas = (time.time() - inicio) * 1000  # Convertido a milisegundos

print(f"Tiempo Python Nativo: {tiempo_python:.2f} ms")
print(f"Tiempo con Pandas:    {tiempo_pandas:.2f} ms")
print(f"Pandas fue ~{tiempo_python / tiempo_pandas:.1f} veces más rápido.")


print("\n=== 3. COMPARACIÓN (Sumar +100 a todos los salarios) ===")

# --- Prueba con Python Nativo ---
inicio = time.time()
salarios_aumentados = [row["salario"] + 100 for row in datos_python]
tiempo_python_op = (time.time() - inicio) * 1000  # Convertido a milisegundos

# --- Prueba con Pandas/NumPy (Vectorizado) ---
inicio = time.time()
_ = df["salario"] + 100
tiempo_pandas_op = (time.time() - inicio) * 1000  # Convertido a milisegundos

print(f"Tiempo Python Nativo (Bucle for): {tiempo_python_op:.2f} ms")
print(f"Tiempo Pandas (Vectorizado):      {tiempo_pandas_op:.2f} ms")
print(f"Operación vectorizada fue ~{tiempo_python_op / tiempo_pandas_op:.1f} veces más rápida.")
```
