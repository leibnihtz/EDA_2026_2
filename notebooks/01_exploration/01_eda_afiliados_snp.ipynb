```python
lima = df[df["dpto"] == 21].copy()
```

```python
lima[["sexoact", "genero"]].value_counts(dropna=False)
```

```python
lima["genero"] = lima["sexoact"].map({
    1: "Categoría 1",
    2: "Categoría 2"
})
```

```python
promedio_genero = (
    lima.groupby("genero")["monto_aportes"]
        .mean()
)

promedio_genero
```

```python
import matplotlib.pyplot as plt

promedio_genero.plot(kind="bar")

plt.title("Monto promedio de aportes por género - Puno")
plt.xlabel("Género")
plt.ylabel("Monto promedio de aportes (S/)")
plt.xticks(rotation=0)
plt.show()
```
